```python
import os
import pandas as pd

def count_total_rows(folder_path):

    total_rows = 0
    csv_files = [file for file in os.listdir(folder_path) if file.endswith(".csv")]

    if not csv_files:
        print("No CSV files found in the directory.")
        return 0

    for file in csv_files:
        file_path = os.path.join(folder_path, file)
        try:
            df = pd.read_csv(file_path)
            row_count = len(df)
            total_rows += row_count
            print(f"{file}: {row_count} rows")
        except Exception as e:
            print(f"Error reading {file}: {e}")

    print(f"\n Total Rows Across All CSVs: {total_rows}")
    return total_rows


folder_path = "/Users/Jerry/Desktop/Immigration Documents/Santa Clara University/Winter 2025/Gen AI/Project/"
total = count_total_rows(folder_path)
print(f"Total number of rows in all CSV files: {total}")

```
